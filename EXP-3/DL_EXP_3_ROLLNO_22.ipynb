```python
print("Roll No: 24BAD022")
print("Name: Divyadharshini S")

import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split

# Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

print("Original Training Shape:", x_train.shape)
print("Original Testing Shape:", x_test.shape)

# Resize images
x_train = tf.image.resize(x_train, (64,64)).numpy()
x_test = tf.image.resize(x_test, (64,64)).numpy()

# Normalize pixel values
x_train = x_train.astype("float32")/255.0
x_test = x_test.astype("float32")/255.0

# Split validation set
x_train, x_val, y_train, y_val = train_test_split(
    x_train, y_train,
    test_size=0.2,
    random_state=42
)

print("\nFinal Dataset Split")
print("Training:", x_train.shape)
print("Validation:", x_val.shape)
print("Testing:", x_test.shape)

class_names = [
    'Airplane','Automobile','Bird','Cat','Deer',
    'Dog','Frog','Horse','Ship','Truck'
]

plt.figure(figsize=(10,5))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(x_train[i])
    plt.title(class_names[int(y_train[i])])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("Roll No: 24BAD022")
print("Name: Divyadharshini S")
model = keras.Sequential([
    keras.layers.Conv2D(
        32,(3,3),
        activation="relu",
        input_shape=(64,64,3)
    ),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Conv2D(
        64,(3,3),
        activation="relu"
    ),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Conv2D(
        128,(3,3),
        activation="relu"
    ),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Flatten(),
    keras.layers.Dense(
        128,
        activation="relu"
    ),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(
        10,
        activation="softmax"
    )
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
print("Roll No: 24BAD022")
print("Name: Divyadharshini S")
history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val,y_val),
    epochs=10,
    batch_size=64,
    verbose=1 )

test_loss,test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTesting Accuracy:",round(test_accuracy*100,2),"%")
print("Testing Loss:",round(test_loss,4))

plt.figure(figsize=(8,5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(8,5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()
plt.show()

print("\nFinal Performance Summary")
print("-"*65)
print(f"{'Metric':<25}{'Value'}")
print("-"*65)
print(f"{'Training Accuracy':<25}{history.history['accuracy'][-1]:.4f}")
print(f"{'Validation Accuracy':<25}{history.history['val_accuracy'][-1]:.4f}")
print(f"{'Training Loss':<25}{history.history['loss'][-1]:.4f}")
print(f"{'Validation Loss':<25}{history.history['val_loss'][-1]:.4f}")
print(f"{'Testing Accuracy':<25}{test_accuracy:.4f}")
print("-"*65)
```

```python
print("Roll No: 24BAD022")
print("Name: Divyadharshini S")
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
predictions = model.predict(x_test,verbose=0)
predicted_classes = np.argmax(predictions,axis=1)
true_classes = y_test.flatten()
cm = confusion_matrix(
    true_classes,
    predicted_classes)
plt.figure(figsize=(8,8))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names)
disp.plot(
    cmap="Blues",
    xticks_rotation=45)
plt.title("Confusion Matrix")
plt.show()

plt.figure(figsize=(12,8))
for i in range(12):
    plt.subplot(3,4,i+1)
    plt.imshow(x_test[i])
    plt.title(
        f"T:{class_names[true_classes[i]]}\nP:{class_names[predicted_classes[i]]}"
    )
    plt.axis("off")
plt.tight_layout()
plt.show()

misclassified = np.where(
    predicted_classes!=true_classes
)[0]
plt.figure(figsize=(12,8))
for i,index in enumerate(misclassified[:12]):
    plt.subplot(3,4,i+1)
    plt.imshow(x_test[index])
    plt.title(
        f"T:{class_names[true_classes[index]]}\nP:{class_names[predicted_classes[index]]}",
        fontsize=9
    )
    plt.axis("off")
plt.tight_layout()
plt.show()
```
